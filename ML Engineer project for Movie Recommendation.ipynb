```python
from ast import literal_eval

def convert(obj):
    L = []
    for i in literal_eval(obj):
        L.append(i["name"])
    return L

df["genres"] = df["genres"].apply(convert)
df["keywords"] = df["keywords"].apply(convert)
```

```python
# recommend system Function with if movie not match and with suggestion 

from difflib import get_close_matches


def recommend(movie):
    matches = df[df["title"].str.lower() == movie.lower()]

    if matches.empty:
        suggestions = get_close_matches(
            movie,
            df["title"].tolist(),
            n=5,
            cutoff=0.5
        )

        print("Movie not found.")
        print("Did you mean:", suggestions)
        return

    index = matches.index[0]
    distances = cosine[index]

    movie_list = sorted(
        list(enumerate(distances)),
        reverse=True,
        key=lambda x: x[1]
    )[1:6]

    print(f"\nMovies similar to {matches.iloc[0]['title']}:\n")

    for i, score in movie_list:
        print(f"{df.iloc[i]['title']}  → similarity: {score:.3f}")
```

```python
# Creating Api 

from fastapi import FastAPI
import pickle

app = FastAPI(
    title="Movie Recommendation API",
    description="Content-based movie recommendation system",
    version="1.0"
)


# Load model artifacts
with open("model/similarity.pkl", "rb") as f:
    similarity = pickle.load(f)

with open("model/movies.pkl", "rb") as f:
    movies = pickle.load(f)


@app.get("/")
def home():
    return {
        "message": "Movie Recommendation API is running"
    }


@app.get("/recommend/{movie}")
def recommend(movie: str):

    # Find movie
    matches = movies[
        movies["title"].str.lower() == movie.lower()
    ]

    if matches.empty:
        return {
            "error": f"Movie '{movie}' not found"
        }

    # Get movie index
    index = matches.index[0]

    # Get similarity scores
    distances = similarity[index]

    # Sort movies by similarity
    movie_list = sorted(
        list(enumerate(distances)),
        reverse=True,
        key=lambda x: x[1]
    )[1:6]

    recommendations = []

    for i, score in movie_list:

        recommendations.append({
            "title": movies.iloc[i]["title"],
            "similarity": round(float(score), 3)
        })

    return {
        "movie": matches.iloc[0]["title"],
        "recommendations": recommendations
    }
```

```python
#improving API 

from fastapi import FastAPI, HTTPException
from pydantic import BaseModel
import pickle


# --------------------------------------------------
# 1. Create FastAPI application
# --------------------------------------------------

app = FastAPI(
    title="Movie Recommendation API",
    description="Content-based movie recommendation system",
    version="1.0.0"
)


# --------------------------------------------------
# 2. Load trained artifacts
# --------------------------------------------------

with open("model/similarity.pkl", "rb") as f:
    similarity = pickle.load(f)

with open("model/movies.pkl", "rb") as f:
    movies = pickle.load(f)


# --------------------------------------------------
# 3. Response schema
# --------------------------------------------------

class Recommendation(BaseModel):
    title: str
    similarity: float


class RecommendationResponse(BaseModel):
    movie: str
    recommendations: list[Recommendation]


# --------------------------------------------------
# 4. Health check
# --------------------------------------------------

@app.get("/")
def home():
    return {
        "status": "success",
        "message": "Movie Recommendation API is running"
    }


# --------------------------------------------------
# 5. Recommendation endpoint
# --------------------------------------------------

@app.get(
    "/recommend/{movie}",
    response_model=RecommendationResponse
)
def recommend(movie: str):

    # Find movie
    matches = movies[
        movies["title"].str.lower() == movie.lower()
    ]

    # Movie not found
    if matches.empty:
        raise HTTPException(
            status_code=404,
            detail=f"Movie '{movie}' not found"
        )

    # Get movie index
    index = matches.index[0]

    # Similarity scores
    distances = similarity[index]

    # Sort by similarity
    movie_list = sorted(
        enumerate(distances),
        key=lambda x: x[1],
        reverse=True
    )[1:6]

    recommendations = []

    for i, score in movie_list:

        recommendations.append(
            Recommendation(
                title=movies.iloc[i]["title"],
                similarity=round(float(score), 3)
            )
        )

    return RecommendationResponse(
        movie=matches.iloc[0]["title"],
        recommendations=recommendations
    )
```
